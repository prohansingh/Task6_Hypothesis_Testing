```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

from scipy.stats import ttest_ind
```

```python
# Create Images Folder

os.makedirs("Task6 images", exist_ok=True)
```

```python
# Load Dataset-

df = pd.read_csv("C:/Users/dell/Documents/amazon_sales_data 2025.csv")

print("Dataset Loaded Successfully")
```

```python
# Basic Cleaning

df = df.drop_duplicates()

# Fill Missing Values
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Select Two Categories

categories = df['Category'].unique()

category1 = categories[0]
category2 = categories[1]

# Extract Sales Data
sales1 = df[df['Category'] == category1]['Total Sales']
sales2 = df[df['Category'] == category2]['Total Sales']
```

```python
# Perform T-Test

t_stat, p_value = ttest_ind(sales1, sales2)
```

```python
# Print Results

print("\nHypothesis Testing Results\n")

print(f"Category 1: {category1}")
print(f"Category 2: {category2}")

print(f"\nT-Statistic : {t_stat}")
print(f"P-Value     : {p_value}")
```

```python
# Interpretation

alpha = 0.05

print("\nInterpretation:\n")

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There is a significant difference between the categories.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is no significant difference between the categories.")
```

```python
# Visualization

plt.figure(figsize=(8,5))

sns.boxplot(
    x='Category',
    y='Total Sales',
    data=df[df['Category'].isin([category1, category2])]
)

plt.title("Sales Comparison Between Categories")

plt.tight_layout()

plt.savefig("Task6 images/category_comparison.png")

plt.close()

print("\nHypothesis Testing Completed Successfully")
```
